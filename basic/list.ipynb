# リスト
- 複数のデータや値を順番に並べて1箇所にまとめて管理するための仕組み
- 角括弧を用いてデータを順番(インデックス)で管理
- リスト内一つ一つのデータを要素と呼ぶ
- インデックスを用いてデータを参照、変更することができる
- インデックスは0から開始

```python
koichi_scores = [85, 92, 78]
```

```python
#koichi_scoreというリスト作成
koichi_score = [85, 92, 78]

# 要素を参照
koichi_score[1]
```

```python
# リストの要素を書き換える(置換)
koichi_score[2] = 77

# 変更した要素を参照
koichi_score[2]
```

# リストを操作する関数
|関数|行う処理|
|---|---|
|len()|リストの要素数を取得|
|sum()|リストの要素の合計値を計算|
|max()|リストで最大の要素を取得|
|min()|リストで最小の要素を取得|
※length(=長さ)

```python
koichi_score = [85, 92, 78]

# 要素数を確認
len(koichi_score)
```

```python
koichi_score = [85, 92, 78]

# 合計を確認
sum(koichi_score)
```

```python
koichi_score = [85, 92, 78]

# 最小値を確認
min(koichi_score)
```

# 2次元リスト
- 2次元リストは、Excelの用に縦横にデータが並んだ表データを管理する
- リストを要素に持つリスト
- Excelのように、縦横にデータが並んだ表を表現するのに便利
- List名[行番号][列番号]

```python
# 2次元リストを作成
koichi_score = [ [85, 92, 78],
                 [70, 88, 95],
                [95, 76, 81]]

koichi_score[0][1]
```

```python
koichi_score[2][2] = 100
koichi_score
```

# 2次元リスト：行数の参照、行の取り出し
- 2次元リストの行数は、len()関数で参照できる
- 2次元リストから行単位でリストを取り出すことができる
- len(List名) #2次元リストの行数の参照
- List名[行番号] #2次元リストから行単位でリストの取り出し

```python
# 2次元リストを作成
koichi_score = [ [85, 92, 78],
                 [70, 88, 95],
                [95, 76, 81]]

len(koichi_score)
```

```python
koichi_score[1]
```

# 辞書
- キーとそれに対応する値をペアにしてデータを管理するデータ型
- 辞書の要素に順番はなく、辞書名[キー]のようにキーを指定し対応する値を直接取得
- 辞書変数名[キー名] = 値 #値の追加、変更
- キー名のキーが辞書内に存在しない場合は新しい値が追加される
- すでに存在している場合はそのキーに対応する値に変更される

```python
koichi_score = {'国語':85,'数学':92,'理科':78}
koichi_score['国語']
```

```python
# 理科の値を書き換える
koichi_score['理科'] = 60
koichi_score['理科']
```

```python
# 英語を追加
koichi_score['英語'] = 90
koichi_score['英語']
```

# 辞書のネスト
- 2次元リストはリストを要素に持つリストだったが辞書ののネストは辞書を辞書を値に持つ辞書
- 辞書の値として別の辞書を入れることができる
- 一つのキーで複数のデータを管理できる

```python
student_score = {'佐藤':{'国語':85,'数学':92,'理科':78},
                 '鈴木':{'国語':70,'数学':88,'理科':95}}

# 佐藤さんのテスト結果をすべて確認
student_score['佐藤']
```

```python
# 鈴木さんの国語の点数を取得
student_score['鈴木']['国語']
```

```python
# 生徒の3教科のテストの点数を辞書にまとめて管理するプログラム
student_score = {
    '佐藤':{'国語':85,'数学':92,'理科':78},
    '鈴木':{'国語':70,'数学':88,'理科':95},
    '高橋':{'国語':95,'数学':76,'理科':81}
}

# 佐藤さんの点数データ(内側の辞書)をすべて取得
sato_scores = student_score['佐藤']
print(f'佐藤さんの全教科の点数:{sato_scores}')

# 鈴木さんの数学の点数を取得
suzuki_math_score = student_score['鈴木']['数学']
print(f'鈴木さんの数学の点数:{suzuki_math_score}点')

# 高橋さんの理科の点数を取得
takahashi_science_score = student_score['高橋']['理科']
print(f'高橋さんの理科の点数:{takahashi_science_score}点')
```
