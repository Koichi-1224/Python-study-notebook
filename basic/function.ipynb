# 関数
- 事前に決められた処理を実行するコード
- print()、type()なども関数の一種
- Pythonに用意されている関数(組み込み関数)
|代表的な関数|処理内容|
|---|---|
|print()|入力を文字列として表示|
|type()|入力データの型を出力|
|sum()|リストの合計値を出力|
|int()|整数のint型に返納|
- 戻り値を入れる変数名 = 関数名(引数)
- 引数や戻り値がない関数もある
- 引数がなくても括弧は必須
- 戻り値の代入は必須ではない

```python
input_value = 3.14
output_value = int(input_value)
output_value
```

# 関数の定義方法と使い方
- 組み込みの関数以外にもオリジナルの関数を作成することもできる
- 何度も同じような処理を書かずに済み、メンテナンスも容易
- defは、define(定義する)の略称。
- 定義の括弧、コロン、インデントは省略できない
- 関数を呼び出す前に関数の定義が必要

```python
def greet():
    print('こんにちは')

greet()
```

```python
greet()
greet()
```

# 関数呼出時のプログラムの流れ
- 関数を呼び出すと本流(元のプログラム)から支流に分岐し、処理が終わると元の場所に戻ってくるイメージ
# 関数の引数
- 関数に変数の値を引き渡したい場合に引数を定義する
- 複数定義もできる

```python
def greet(name):
    print(f'{name}さん、こんにちは')

greet('小川')
```

```python
greet('佐藤')
```

# 関数の戻り値
- 関数から処理結果を受け取りたい場合に戻り値(返り値)を定義

```python
def greet(name):
    string = f'{name}さん、こんにちは'
    return string

s = greet('小川')
print(s)
```

# 関数の引数と戻り値
- 関数には、引数と戻り値の両方を指定した処理を定義できる

```python
# 円の面積を計算する関数
def calculate_circle_area(r):
    pi = 3.14
    area = r * r * pi
    return area

r = 10
area = calculate_circle_area(r)
print(f'半径{r}の円の面積は{area}です。')
r = 5
area = calculate_circle_area(r)
print(f'半径{r}の円の面積は{area}です。')
```

# キーワード引数
- どの引数にどの値を渡すかを、明示的にしていすることができる
- 仮引数(関数内で使われている引数名) = 実引数(与える値)のように渡す
- キーワード引数で指定することでそれぞれの値がどの引数に渡されているのかわかりやすくなる
- キーワード引数では順番に関係はなく、指定した仮引数に値が渡される

```python
# 円柱の体積を計算する関数
def calculate_cylinder_volume(r,h):
    pi = 3.14
    volume = r * r * pi * h
    return volume

volume = calculate_cylinder_volume(r=5,h=4)
print(f'円柱の体積は{volume}です')
```

# デフォルト引数
- デフォルト引数を指定することにより引数を省略した場合に使われる値を設定可能
- def文で仮引数 = 初期値のように定義する

```python
# 円柱の体積を計算する関数
def calculate_cylinder_volume(r,h=4):
    pi = 3.14
    volume = r * r * pi * h
    return volume

volume = calculate_cylinder_volume(5)
print(f'円柱の体積は{volume}です')
volume = calculate_cylinder_volume(5,8)
print(f'円柱の体積は{volume}です')
```

# モジュール
- モジュールとは、プログラムを再利用するためのコードをファイルとしてまとめたもの
- 大きなプログラムを複数の小さなプログラムに分割して管理できる
- 他人が開発したプログラムを簡単に利用できる
- importの書式：import モジュール名
- importされた関数の呼び出し方法：モジュール名.関数名()

```python
# functions.pyのモジュールのコード
def nibai(x):
    return 2 * x
def sanbai(x):
    return 3 * x
```

```python
import functions

f = functions.nibai(2)
g = functions.sanbai(2)
h = f + g
print(h)
```

# クラス
- クラスとはデータと処理をまとめたもの
- 例えば、、、
- 表データを管理するクラス
- データを可視化するクラス
- 機械学習モデルのクラス
# クラスとインスタンス
- クラスはデータと処理をまとめた雛形
- クラスを使うためにはそのクラスのインスタンスを生成する必要がある

クラス(属性：　メソッド：)
インスタンス１(属性:### メソッド:###)


# クラスの利用方法
- インスタンスは変数へ代入することで利用可能
- インスタンスを代入した変数によって、属性とメソッドをつかうことができる
## インスタンスの生成方法
- 変数名 = クラス名()
- 関数を使用するときのように括弧は必須
- カッコ内に値を入れる必要がある場合もある
## 属性の参照方法
- クラスを代入した変数名.属性名
- 通常の変数の参照と異なり、クラスの代入した変数名と.(ドット)が必要
## メソッドの使用方法
- クラスを代入した変数名.メソッド名()
- 通常の関数の使用方法と異なり、クラスを代入した変数名と.(ドット)が必要

# 機械学習におけるクラスの利用例
- 線形回帰のクラスでは傾き・切片(属性)を求めるメソッド(fit)を活用する
ex) 線形回帰モデル「y = ax + b」の例

```python
from sklearn.linear_model import LinearRegression

x_data = [[28],[30],[32],[34]]
y_data = [10,80,100,120]

model = LinearRegression() # クラスのインスタンス生成

model.fit(x_data,y_data) # fitメソッドでdataにフィット

print('傾き：',int(model.coef_[0]))
print('切片：',int(model.intercept_))
```
