# 条件分岐(if文)
- 条件分岐(if文)は条件の真偽値によって処理の流れを変更するコードの記載方法

```python
is_rain = True #真のとき
if is_rain:
    print('傘をもつ')
```

Pythonではインデントが文法の一部であり、非常に重要。
：の次の行から実行したい処理のブロックを半角スペース４つでインデントする。

# 条件分岐(if-else文)
- if文にelseを使用すると条件が偽となる場合の処理を付け足せる

```python
is_rain = False
if is_rain:
    print('傘を持つ')
else:
    print('新しいくつをはく')
```

# 比較演算子
|演算子記号|意味|
|---|---|
|==|左右の値が同じ|
|!=|左右の値が同じでない|
|>|左が右の値より大きい|
|<|左が右の値より小さい|
|>=|左が右の値以上|
|>=|左が右の値以上|

```python
print(1 == 1) #=は代入
```

```python
print(1 < 0)
```

```python
age = 18
if age < 20:
    print('No Drink')
```

# 論理演算子
- 複数の演算の真偽値を評価するもの
- if文で比較演算子と組み合わせて使用することが多い

|演算子記号|意味|
|---|---|
|and|かつ|
|or|または|
|not|でなければ|


|演算子|優先度|
|---|---|
|算術|高|
|比較|中|
|論理|低|

```python
print(True and False)
```

```python
print(True or False)
```

```python
writing = 85
listening = 90
if writing >= 70 and listening >= 80:
    print('English:grade A')
```

# 条件分岐(if-elif-else文)
- 条件分岐を３つ以上にしたい場合にelifを追加
- 比較・論理演算子と組み合わせてより複雑な条件分岐が可能

```python
writing = 50
listening = 75
if writing >= 70 and listening >= 80:
    print('English:grade A')
elif writing >= 60 or listening >= 75:
    print('English:grade B')
else:
    print('English:grade C')
```

# 繰り返し(for文)
- for文は同じ処理を何度も書かずに繰り返したい場合に使用
- 範囲をrange()で指定して繰り返せる

```python
for i in range(5): #変数名はよくiかjが使われる。カッコ内の5は省略形。(0,5)とも書くことができる
    print('Pythonの文法')
```

# for文とリスト
- for文にリストを組み合わせると各要素を取り出して同じ処理を繰り返すことができる

```python
menu = ['salad','soup','steak','cake']

for food in menu:
    print(food)
```

# for文と辞書
- for文と辞書を組み合わせると各要素(キー、値)を取り出して同じ処理を繰り返すことができる
- 辞書の要は前から順に代入
- .items()は辞書のキーとそれに対応する値を取得するメソッド
- メソッドとは特定のデータ型に対し決まった処理を実行できるもので、データ型.メソッド名()のように使う

```python
order = {'Taro':'soup',
'Jiro':'steak',
'Hanako':'cake'}
for person,food in order.items():
    print(f'{person}s order is {food}.')
```
