## 変数
- 変数:値に名前を付けて一時的に記録できる箱
- =をつかって箱に値を「代入」、変数名を使って中身を「参照」することができる

```python
# pythonという変数に数値50を代入
python = 50

# 変数を参照
python
```

```python
# pythonという変数に'小川くん'という文字列を代入
python = '小川くん' # 文字列はシングルクオテーションで囲む。ダブルクオーテーションでも同じ。

# 変数名を使って中身を参照
python
```

```python
# 再代入
python = 50
python

python = 100
python
```

# input関数の扱い方
- input()で受け取った値のデータ型はかならず文字列型になる
- 他のデータ型を扱いたい場合は型変換を行う必要がある

```python
kokugo = int(input('国語の点数を入力してください: '))
sugaku = int(input('数学の点数を入力してください: '))
kokusu = kokugo + sugaku
print(f'国語と数学の点数: {kokusu}')
```

```python
# ====生徒の名前を入力する===
seito = input('生徒の名前を入力してください:')

# ===3教科の点数を入力する===
kokugo = int(input('国語の点数を入力してください: '))
sugaku = int(input('数学の点数を入力してください: '))
eigo = int(input('英語の点数を入力してください: '))

# ===合計点を計算する===
total = kokugo + sugaku + eigo

# ===平均点を計算する===
average = total / 3

# ===結果を表示する=== 
print('---') #区切り線
print(seito+'の点数')
print('3教科合計点は'+str(total)+'点です')
print(f'平均点は{average}点です')
```
