## for + if：真夏日判定

最高気温 `[28, 32, 35, 30, 25]`。30度以上が真夏日。

### Q3-22  各日の `'真夏日'` / `'通常'` を順に並べたリスト `labels` に

```python
temps = [28, 32, 35, 30, 25]

labels = []
for i in temps:
    if i >= 30:
        labels.append('真夏日')
    else:
        labels.append('通常')

check('3-22', labels)
answer('3-22')
```

### Q3-23  真夏日の日数を `hot_count` に

```python
hot_count = 0
for i in temps:
    if i >= 30:
        hot_count = hot_count + 1

check('3-23', hot_count)
answer('3-23')
```

## 総合：勉強時間

| 月 | 火 | 水 | 木 | 金 | 土 | 日 |
|---|---|---|---|---|---|---|
| 2 | 0 | 3 | 1 | 2 | 4 | 0 |

2時間以上を「ちゃんとやった日」とする。

### Q3-24  表を辞書 `study` に（キーは `'月'` のような1文字）

```python
study = {'月': 2 , '火': 0 , '水': 3 , '木': 1 , '金': 2 , '土': 4 ,'日': 0}

check('3-24', study)
```

### Q3-25  ちゃんとやった日数を `good_count` に

```python
good_count = 0
for day, hours in study.items():
    if hours >= 2:
        good_count = good_count + 1

check('3-25', good_count)
answer('3-25')
```

### Q3-26  ちゃんとやった曜日のリストを `good_days` に

```python
good_days = []
for day,hour in study.items():
    if hour >= 2:
        good_days.append(day)

check('3-26', good_days)
answer('3-26')
```

### Q3-27  【ちょい難】一番勉強した曜日を `best_day` に（`max()` は使わず for と if で）

```python
best_day = ''
best_hours = 0
for day,hours in study.items():
    if hours > best_hours:
        best_hours = hours
        best_day = day
        
check('3-27', best_day)
answer('3-27')
```
