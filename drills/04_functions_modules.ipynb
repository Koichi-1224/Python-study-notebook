## 関数の定義

### Q4-1  正なら `'plus'`、負なら `'minus'`、0なら `'zero'` を返す `sign(x)` を定義

```python
def sign(x):
    if x > 0:
        return 'plus'
    elif x < 0:
        return 'minus'
    else:
        return 0

check('4-1', sign(-5))
```

### Q4-2  長方形の面積を返す `area(width, height)` を定義

```python
def area(width, hight):
    return width * hight

check('4-2', area(4, 5))
```

### Q4-3  【デフォルト引数】`area` を、`width` も `height` も省略したら 1 になるように定義し直す

```python
def area(width=1, height=1):
    return width * height

check('4-3', area(height=7))
```

### Q4-6  BMI = 体重(kg) ÷ 身長(m)² を返す `bmi(weight, height_m)` を定義

```python
def bmi(weight, height_m):
    return weight / height_m ** 2

check('4-6', bmi(60, 1.7))
```

### Q4-7  Q4-6 の `bmi` を中で呼んで、18.5未満 → `'やせ'`、25未満 → `'普通'`、それ以外 → `'肥満'` を返す `bmi_label(weight, height_m)` を定義

```python
def bmi_label(weight, height_m):
    b = bmi(weight,height_m)
    if b < 18.5:
        return 'やせ'
    elif b < 25:
        return '普通'
    else:
        return '肥満'

check('4-7', bmi_label(60, 1.7))
```
