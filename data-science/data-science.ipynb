# 機械学習モデル「線形回帰」
線形回帰とは、予測するための変数である説明変数と、予測したい変数である目的変数の関係を直線で表現する手法。<br>
説明変数を入手したときに得られる目的変数の予測値が、実際の目的変数の値に近づくようにモデルを学習させる。説明変数が1つの場合は単回帰分析、説明変数が複数の場合は重回帰分析と呼ばれる。<br>
単回帰分析はyを目的変数、xを説明変数をした時、2つのデータ間の関係性を
```
\begin{eqnarray}
y = ax + b
\end{eqnarray}
```
という一次関数の形で表す。<br>
このとき、係数aと切片bが分かれば、xを代入することで、yを予測することができる。

重回帰分析は、yは目的変数、x1,x2,・・・xnを説明変数としたとき、複数のデータ間の関係性を次のような一次関数の形で表す。
```
\begin{eqnarray}
y = a_1 \cdot x_1 + a_2 \cdot x_2 + \cdots + a_n \cdot x_n + b
\end{eqnarray}
```
ここで、a1,a2,・・・anは各説明変数に対応する係数(「回帰係数」)であり、bは切片。これらの係数と切片が分かれば、各説明変数x1,x2,・・・xnまでのn個の値を代入することで、yを予測することができる。

# モデルの構築
モデルの構築には、機械学習のライブラリである「scikit-learn」を使用する。このライブラリには、さまざまな機械学習モデルが用意されており、簡単に利用できるようになっている。<br>
インポート方法は、pandasとほぼ同じ。使用したいモデルが明確に決まっている場合、そのモデルだけを指定してインポートできる。なお、このライブラリの正式な名前は「scikit-laern」だが、import文では「sklearn」と記述する。

この場合、sklearnの線形モデル関連のlinear_modelモジュールから、線形回帰用のクラスであるLinearRegressionをインポートして使用する。
```
[scikit-learnのlinear_modelモジュールからLinearRegressionクラスをインポート]
from sklearn.linear_model import LinearRegression
```
fromの後にライブラリ名とモジュール名を指定し、その中からLinearRegressionクラスだけをインポートしている。なお、linear_modelモジュールには他にもさまざまなクラスが含まれているが、このように必要なクラスだけをインポートすることで、使用する際にも「モジュール名.クラス名」とせずにクラス名だけで使用することができる。

````python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
````

この2行だけで線形回帰のモデルを作成できる！<br>
LinearRegressionクラスをインポートした後に、二行目ではクラスをインスタンス化している。(クラスから実際に使える"モデルの実態"を作ることをインスタンス化と呼ぶ)<br>
このように、LinearRegressionというクラス名の後にかっこを付け、それを変数に代入することで、モデルのインスタンス化が簡単に行える。必要なクラスだけをインポートしているので、クラス名だけで使用できている。<br>
あとは実際にデータをこのモデルに学習させるだけで、予測モデルが完成する。

````python
import pandas as pd 
test_data = pd.DataFrame([[65,120,12000],[70,115,15000],[66,130,13500],[64,133,12100],[67,127,14200]], columns = ['width', 'engine-size', 'price'])
````

次にこの表から説明変数と目的変数を分けて、機械学習モデルに学習させよう。特定の列だけを抜き出すには、列名をリストとして指定<br>
[['列名1','列名2']]のように抜き出したい列をカンマで区切ってリストにまとめ、それをさらに角かっこで囲むこてで、指定した列を抜き出すことができる。

````python
x = test_data[['width', 'engine-size']]
x
````

目的変数tも同様に抜き出しますが、1列だけでよいので、リストではなく列名だけをそのまま指定する。今回は'price'

````python
t = test_data['price']
t
````

`scikit-learn`の機械学習モデルの学習には`fit`メソッドを利用する。

````python
from sklearn import set_config
set_config(display='text')

model.fit(x, t)
````

今回は説明変数が'width'と'engine-size'の2つなので、
```
\begin{eqnarray}
y = a_1 \cdot x_1 + a_2 \cdot x_2  + b
\end{eqnarray}
```
という一次関数を想定している。

指定された切片は`intercept_`という属性に、係数は`coef_`という属性に格納させる。文字列では中かっこ(波かっこ)「{}」の中に変数名を書き、文字列全体の前にfを付けることで、変数の値を文字列の中に直接埋め込むことができる

````python
print(f'推定された切片: {model.intercept_}')
print(f'推定された係数: {model.coef_}')
````

ここで回帰係数$a$や切片$b$の推定には**最小二乗法**という手法が用いられる。<br>
これは、実際の目的変数の値(実測値)と予測値の差の二乗和を計算し、それが最小になるときの回帰係数と切片を求める方法。<br>
以下は説明変数が$m$個ある場合の、$i$番目のデータに対する予測値を表す式<br>
$$
f(x_i) = a_1 x_{i1} + a_2 x_{i2} + \dots + a_m x_{im} + b
$$
最小二乗法では、$y_i$を実測値として以下の式の値が最小になるように計算する
$$
\sum_{i=1}^{n} \left( y_i - f(x_i) \right)^2
$$

- モデルのインスタンス化 -> model = LinearRegression()
- データへのフィット -> model.fit()
