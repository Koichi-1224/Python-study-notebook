# 1.データの理解

```python
import pandas as pd
```

```python
df = pd.DataFrame([[64,30,7],[67,54,15],[74,51,12],[46,31,19],[72,62,16],[75,65,15],[68,57,16],[65,42,16],[58,42,19],[73,65,12]])
df
```

このデータフレームdfは、ある10人の生徒のデータ
- 0:数学のテストの点数(100点満点)
- 1:勉強時間(1週間あたり)
- 2:ゲーム・SNS等スマートフォンの利用時間(1週間あたり) <br>
今回は、勉強時間とスマートフォンの利用時間から数学のテストの点数を予測することを目的とする

```python
df.columns= ['math_scores', 'study_hours', 'smartphone_hours']
```

# 2.モデルの構築と評価

```python
# モデル構築準備
x = df[['study_hours', 'smartphone_hours']] # 説明変数
t = df['math_scores'] # 目的変数
```

```python
# モデル構築
from sklearn.linear_model import LinearRegression
model = LinearRegression()

model.fit(x,t)
print(f'切片:{model.intercept_}')
print(f'係数:{model.coef_}')
```
